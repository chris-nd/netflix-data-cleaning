# Nettoyage des données

Application de la logique de nettoyage aux données conformement aux décisions prises dans les notes d'exploration.

```python
import pandas as pd

from cleaning_netflix.pipeline import clean_netflix
from cleaning_netflix.storage import load_raw, save_clean
```

## Chargement des données

```python
netflix_raw = load_raw()
```

## Vérifications avant nettoyage

Du nombre de lignes, colonnes, nombre de NaN par colonne 

```python
print(netflix_raw.shape)
netflix_raw.isna().sum()
```

## Nettoyage des données

```python
netflix_clean = clean_netflix(netflix_raw)
```

## Vérifications après nettoyage

Usage des `assert` explicite pour vérifier que les données sont correctement nettoyées

```python
assert len(netflix_clean) == len(netflix_raw)
assert "duration" not in netflix_clean.columns
assert netflix_clean["show_id"].is_unique

structural = [
    "date_added",
    "month_added",
    "year_added",
    "duration_minutes",
    "duration_seasons",
]

assert netflix_clean.drop(columns=structural).isna().sum().sum() == 0

assert (
    netflix_clean["date_added"].isna().sum() == netflix_raw["date_added"].isna().sum()
)
assert (
    netflix_clean["duration_minutes"].notna().equals(netflix_clean["type"] == "Movie")
)
assert (
    netflix_clean["duration_seasons"].notna().equals(netflix_clean["type"] == "TV Show")
)
```

## Enregistrement des données nettoyées

```python
save_clean(netflix_clean)
```
