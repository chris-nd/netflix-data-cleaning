# Nettoyage des données

Application de la logique de nettoyage aux données conformement aux décisions prises dans les notes d'exploration.

```python
import numpy as np
import pandas as pd
```

```python
netflix = pd.read_csv('../data/raw/netflix_titles_raw.csv')
```

## Gestion des valeurs manquantes

### Colonne `director`

```python
netflix['director'] = netflix['director'].fillna('Not Specified')
```

### Colonne `cast`

```python
netflix['cast'] = netflix['cast'].fillna('Not Specified')
```

### Colonne `country`

```python
netflix['country'] = netflix['country'].fillna('Not Specified')
```

### Colonne `duration`

```python
mask_duration = netflix['duration'].isna()
netflix.loc[mask_duration, 'duration'] = netflix.loc[mask_duration, 'rating']
netflix.loc[mask_duration, 'rating'] = np.nan
netflix.loc[mask_duration]
```

### Colonne `rating`

```python
netflix['rating'] = netflix['rating'].fillna("Unknown")
```
